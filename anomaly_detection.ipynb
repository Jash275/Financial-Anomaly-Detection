# Financial Transaction Anomaly Detection

Using an Isolation Forest model to flag statistically anomalous transactions in a dataset of 217,441 real-world financial transactions.

**Dataset:** Kaggle — Financial Anomaly Data  
**Model:** Isolation Forest (unsupervised outlier detection)

```python
import pandas as pd
from sklearn.ensemble import IsolationForest

# Load data
df = pd.read_csv('financial_anomaly_data.csv')
print(df.shape)
print(df.head())
```

```python
# Confirm column names
print(df.columns.tolist())
```

## Anomaly Detection

An Isolation Forest is trained on transaction amounts. Unlike fixed-rule thresholds (e.g. "flag anything above ₹50,000"), the model learns what a 'normal' transaction amount looks like from the data itself and isolates statistical outliers — catching risk patterns a manual rule would miss.

`contamination=0.02` tells the model to expect roughly 2% of the data to be anomalous.

```python
model = IsolationForest(contamination=0.02, random_state=42)
df['anomaly'] = model.fit_predict(df[['Amount']])

# -1 = anomaly (flagged as high-risk), 1 = normal
flagged = df[df['anomaly'] == -1]
print(f"Total flagged transactions: {len(flagged)}")
print(flagged.head(20))
```

```python
# Save flagged transactions for downstream reporting (Excel VBA + Power BI)
flagged.to_csv('flagged_transactions.csv', index=False)
print("Saved successfully!")
```

## Results Summary

- **Total transactions analyzed:** 217,441
- **Flagged as anomalous:** 4,345 (≈2% of total volume)
- **Total flagged exposure:** ₹176.21M

The `flagged_transactions.csv` output feeds into the Excel VBA automated reporting step and the Power BI dashboard.
